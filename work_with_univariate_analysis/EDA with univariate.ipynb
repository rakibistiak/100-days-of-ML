```python
import matplotlib.pyplot as plt

# Visualize survival counts
sns.countplot(x='Survived', data=df)
plt.title("Titanic Survival Counts (0 = Died, 1 = Survived)")
plt.show()
```

```python
df['Pclass'].value_counts()
```

```python
#CountPlot for Categorical Value
ax=sns.countplot(x='Pclass', data=df, hue='Survived')
for p in ax.patches:
    count=p.get_height()
    ax.annotate(f'{count}',
                (p.get_x()+p.get_width()/2,p.get_height()), va='center',ha='center',
        
    )
```
