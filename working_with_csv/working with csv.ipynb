<h2>Converters</h2>
<b> If we want to rename any cols data we can do that like cols data has Bangladesh Cricket Board we can write it to BCB</b>

```python
def rename(name):
    if name=='Royal Challengers Bangalore':
        return 'RCB'
    elif name=='Mumbai Indians':
        return 'MI'
    elif  name=='Rajasthan Royals':
        return 'RR'
    elif  name=='Kings XI Punjab':
        return 'KXP'
    elif  name=='Kolkata Knight Riders':
        return 'KKR'
    else:
        return name
```

```python
rename('Kolkata Knight Riders')
```

```python
pd.read_csv('IPL Matches 2008-2020.csv', converters={'team1':rename, 'team2':rename})
```
